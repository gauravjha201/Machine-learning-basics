```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
x,y=load_iris(return_X_y=True)
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
x_train
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
gnb.fit(x_train,y_train)
```

```python
y_pred=gnb.predict(x_test)
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

```python
x=df[['total_bill','tip','sex','day','time','size']]
y=df['smoker']
x,y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Re-initialize train/test splits to ensure x_train, x_test, y_train, y_test are fresh categorical strings
x_train,x_test,y_train,y_test=train_test_split(x.copy(),y.copy(),test_size=0.3,random_state=10)

# LabelEncoder for the target variable 'smoker' (y)
lb3=LabelEncoder()
lb3.fit(y.astype(str))

y_train=lb3.transform(y_train.astype(str))
```

```python
# Only transform y_test here, as x_test feature encoding will be handled by ColumnTransformer
y_test=lb3.transform(y_test.astype(str))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Define the ColumnTransformer to OneHotEncode 'sex', 'day', 'time' and passthrough the rest.
# Column indices: 'total_bill'(0), 'tip'(1), 'sex'(2), 'day'(3), 'time'(4), 'size'(5)
ct=ColumnTransformer(
    transformers=[('onehot', OneHotEncoder(drop='first'), [2, 3, 4])],
    remainder='passthrough'
)

import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)

x_train=ct.fit_transform(x_train)
```

```python
x_test=ct.transform(x_test)
```

```python
x_test
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
gnb.fit(x_train,y_train)
```

```python
y_pred=gnb.predict(x_test)
from sklearn.metrics import accuracy_score,classification_report
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
