KNN Classifier

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
x,y=make_classification(n_samples=1000,n_features=3,n_redundant=1,n_classes=2,random_state=999)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier

classifier=KNeighborsClassifier(n_neighbors=5,algorithm='auto')
classifier.fit(x_train,y_train)
```

```python
y_pred=classifier.predict(x_test)
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
best_accuracy = 0
for k in range(1, 11):
  current_classifier = KNeighborsClassifier(n_neighbors=k, algorithm='auto')
  current_classifier.fit(x_train, y_train)
  y_pred_k = current_classifier.predict(x_test)
  current_accuracy = accuracy_score(y_test, y_pred_k)
  best_accuracy = max(best_accuracy, current_accuracy)
print(best_accuracy)
```

KNN Regression

```python
from sklearn.datasets import make_regression
x,y=make_regression(n_samples=1000,n_features=2,noise=10,random_state=42)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsRegressor
regression=KNeighborsRegressor(n_neighbors=5,algorithm='auto')
regression.fit(x_train,y_train)
```

```python
y_pred=regression.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```
